#### RDBMS
       RDBMS is a type of software which stores the data in neat, organised table with rows and columns. The tables are connected through via primary key and foreign keys.

- Database is a collection of objects.
- Table is a collection of records.
- Records are collection of field or attributes.
- All fields or attributes has its own datatypes.


### TABLE:
       table is a structured collection of data organised as a grid of horizontal rows and vertical columns.
