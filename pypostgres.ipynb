```python
SELECT * FROM students;
SELECT * FROM courses;
SELECT * FROM student_courses;
 
SELECT
    s.student_id,
	s.name
    s.name student_name,
	sc.enrollment_date,
	sc.course_id,
    c.course_code,
    c.course_name
FROM students AS s
JOIN student_courses AS sc
ON s.student_id = sc.student_id
JOIN courses AS c
ON sc.course_id = c.course_id

SELECT
     t.teacher_id,
	 t.name,
	 c.course_name
FROM teachers as t
JOIN courses AS c
ON t.teacher_id = c.teacher_id
WHERE t.teacher_id = 1
ORDER BY t.name desc, c.course_name

SELECT * FROM students;
SELECT * FROM courses;
SELECT * FROM student_courses;
SELECT 
s.student_id,
s.name,
c.course_code,
c.course_name
FROM students as s
join student_courses as sc
on s.student_id = sc.student_id
JOIN courses as c
ON sc.course_id = c.course_id


select * from courses;
select * from teachers;
select
c.course_id,
t.name,
c.course_name,
t.department
from courses as c
join teachers as t
on c.teacher_id = t.teacher_id;


--Show student marks
select * from marks;
select * from students;
select * from courses;
select 
s.name,
c.course_code,
c.course_name,
m.marks
from marks m
join students s
on m.student_id = s.student_id
join courses c
on m.course_id = c.course_id
order by marks desc;

--Show students, courses, teachers and marks together

select 
s.name,
c.course_code,
c.course_name,
t.name,
m.marks
from marks m
join courses c
on m.course_id = c.course_id
join students s
on m.student_id = s.student_id
join teachers t
on c.teacher_id = t.teacher_id
order by s.name, c.course_name;

--Find students who scored above 80
```
